### 標本サイズにおける base_information.csv の出力実験

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os 
from pathlib import Path
```

```python
dir_path = Path('../sample_data')
file_paths = sorted([str(p) for p in dir_path.rglob("*") if p.is_file()])
```

```python
for file_path in file_paths[:17]:
    print(f"\r File: {file_path}", end="", flush=True)
```

```python
base_information = {
    "date": [],
    "name": [],
    "cols": [],
    "rows": [],
}
```

```python
for file_path in file_paths:
    try :
        df = pd.read_csv(file_path)
        cols = len(df.columns)
        rows = len(df)
    except Exception as e:
        print(f"\rError reading {file_path}: {e}".ljust(110), end="")
        cols = None
        rows = None
        continue
    finally:
        date = file_path.split('\\')[-2]
        name = os.path.basename(file_path)

        base_information["date"].append(date)
        base_information["name"].append(name)
        base_information["cols"].append(cols)
        base_information["rows"].append(rows)
```

```python
base_information_df = pd.DataFrame.from_dict(base_information)
```

```python
base_information_df[base_information_df['name'] == 'derivatives_bars_daily_futures.csv'].value_counts()
```

```python
name_list = [
    "derivatives_bars_daily_futures.csv",
    "derivatives_bars_daily_options.csv",
    "derivatives_bars_daily_options_225.csv",
    "equities_bars_daily.csv",
    "equities_earnings-calendar.csv",
    "equities_investor-types.csv",
    "equities_master.csv",
    "fins_details.csv",
    "fins_dividend.csv",
    "fins_summary.csv",
    "indices_bars_daily.csv",
    "indices_bars_daily_topix.csv",
    "markets_breakdown.csv",
    "markets_calendar.csv",
    "markets_margin-alert.csv",
    "markets_margin-interest.csv",
    "markets_short-ratio.csv"
]
```

```python
col_dict = {}
for name in name_list:
    cols_set = base_information_df[base_information_df['name'] == name]['cols'].dropna().to_numpy()
    rows_set = base_information_df[base_information_df['name'] == name]['rows'].dropna().to_numpy()
    # rows_set = set(base_information_df[base_information_df['name'] == name]['rows'])
    rows_set_mean = base_information_df[base_information_df['name'] == name]['rows'].dropna().mean()
    rows_set_std = base_information_df[base_information_df['name'] == name]['rows'].dropna().std()
    col_dict[name] = (cols_set, rows_set, rows_set_mean, rows_set_std)

col_df = pd.DataFrame.from_dict(col_dict, orient='index', columns=['cols_num', 'rows_num', 'rows_set_mean', 'rows_set_std'])
col_df.reset_index(inplace=True)
```

### 本番サイズにおける base_information.csv の出力実験
任意の入力フォルダから任意の出力フォルダへ基礎情報を出力

```python
input_datadir_name = "data_financial"
output_datadir_name = "detail_base_information"
```

```python
name_list = [
    "derivatives_bars_daily_futures.csv",
    "derivatives_bars_daily_options.csv",
    "derivatives_bars_daily_options_225.csv",
    "equities_bars_daily.csv",
    "equities_earnings-calendar.csv",
    "equities_investor-types.csv",
    "equities_master.csv",
    "fins_details.csv",
    "fins_dividend.csv",
    "fins_summary.csv",
    "indices_bars_daily.csv",
    "indices_bars_daily_topix.csv",
    "markets_breakdown.csv",
    "markets_calendar.csv",
    "markets_margin-alert.csv",
    "markets_margin-interest.csv",
    "markets_short-ratio.csv"
]
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os 
from pathlib import Path
```

```python
dir_path = Path(f'../{input_datadir_name}')
file_paths = sorted([str(p) for p in dir_path.rglob("*") if p.is_file()])
```

```python
base_information = {
    "date": [],
    "name": [],
    "cols": [],
    "rows": [],
}
```

```python
for file_path in file_paths:
    try :
        df = pd.read_csv(file_path)
        cols = len(df.columns)
        rows = len(df)
    except Exception as e:
        print('\r', f"Error reading {file_path}".ljust(110, '-') + f": {e}", end="")
        cols = None
        rows = None
        continue
    finally:
        date = file_path.split('\\')[-2]
        name = os.path.basename(file_path)

        base_information["date"].append(date)
        base_information["name"].append(name)
        base_information["cols"].append(cols)
        base_information["rows"].append(rows)
```

```python
base_information_df = pd.DataFrame.from_dict(base_information)
```

```python
base_information_df[base_information_df['name'] == 'derivatives_bars_daily_futures.csv'].value_counts()
```

```python
now_time = pd.Timestamp.now().strftime("%Y-%m-%d %H-%M-%S")
os.makedirs(f'../data_output/{now_time}', exist_ok=True)
base_information_df.to_csv(f'../data_output/{now_time}/base_information.csv', index=False)

os.makedirs(f'../data_output/{now_time}/{output_datadir_name}', exist_ok=True)
for i, name in enumerate(name_list): 
    dict_tmp = {key: value[i::len(name_list)] for key, value in base_information.items()}
    pd.DataFrame.from_dict(dict_tmp).to_csv(f'../data_output/{now_time}/{output_datadir_name}/{name}', index=False)
print("Done!")
```
