```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python
### all paths

nparray_dir = Path("../masks")
data_dir = Path("../data_financial")

### check if paths exist

exist = Path.exists(nparray_dir) and Path.exists(data_dir)
print(f"Exist: {exist}")
```

```python
### define variables

experiment_date = "2008-05-07"
```

```python
date_array: np.ndarray = np.load(nparray_dir / "only_date_nparray.npy", allow_pickle=True)
print(f"TYPE  : {type(date_array)}")
print(f"Shape : {date_array.shape}")
print(f"Values: {date_array}")
```

```python
### TEST, on 2008-05-07 

df_dict = {
    "name":[],
    "path":[],
    "df":[],
    "shape":[]
}
for i, path in enumerate((data_dir / experiment_date).glob("*.csv")):
    try:
        tmp_df = pd.read_csv(path)
        shape = tmp_df.shape
    except Exception as e:
        shape = (0, 0)
    finally:
        print(str(i).rjust(2), "|", str(path.name).ljust(38), "|", shape)
        df_dict["name"].append(str(path.name))
        df_dict["path"].append(path)
        df_dict["df"].append(tmp_df)
        df_dict["shape"].append(shape)
```

concat <strong>equities_bars_daily</strong> + <strong>equities_master</strong> by date column

```python
equities_bars_daily: pd.DataFrame = df_dict["df"][3]
equities_master: pd.DataFrame = df_dict["df"][6]
```

```python
### display both of dataframes

display(equities_bars_daily)
display(equities_master)
```

```python
### check both Code columns are the same

code_3 = equities_bars_daily["Code"]
code_6 = equities_master["Code"]
if code_3.equals(code_6):
    print("Code columns are the same.")
```

```python
### concat both dataframes

concat_df = equities_master.merge(equities_bars_daily, on=["Date", "Code"], suffixes=("_1", "_2"))
print(f"Concatenated DataFrame shape: {concat_df.shape}")
```

```python
concat_df[concat_df.isna().any(axis=1)].to_csv("tmp.csv", index=False)
```

```python
concat_df.to_parquet("concat_df.parquet", index=False)
```
