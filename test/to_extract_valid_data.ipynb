```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import shutil
from pathlib import Path
```

## Description

このipynbファイルは，フォルダ構造を整理するために作成しています．

全ての取得可能な日付についてデータを取得後，ターゲットノードとなる「equities_bars_daily」の有効な日付に従ってデータを取捨しています．

このとき，evacuated_data以下に移動したファイルは不要なデータとみなした．

```python
extract_date_array:np.ndarray = np.load('../masks/only_date_nparray.npy', allow_pickle=True)
print(extract_date_array)
```

```python
parent_dir = Path("../all_data/")
dest = Path("../evacuated_data/")
dest.mkdir(exist_ok=True)
```

```python
for subdir in parent_dir.iterdir():
    if subdir.is_dir():
        if subdir.name not in extract_date_array:
            shutil.move(str(subdir), str(dest / subdir.name))
            print(f"\r Moved {subdir} to {dest / subdir.name}", end="")
```
