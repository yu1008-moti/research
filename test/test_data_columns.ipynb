```python
csv_dir = Path('../csv')
sampling_date_range = pd.date_range(start='2023-01-01', end='2023-12-31', freq='D')
```

```python
def get_file_paths(date: pd.Timestamp) -> List[str]:
    date_str = date.strftime('%Y-%m-%d')
    sampled_data_dir = csv_dir / date_str
    return [str(p) for p in sampled_data_dir.rglob("*") if p.is_file()]
```
